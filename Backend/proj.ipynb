```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('english'))


def clean_text(text):
    text = text.lower()
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'https?://\S+|www\.\S+|\S+\.\S{2,}', "", text)
    # Remove stopwords
    text = ' '.join([word for word in text.split() if word not in stop_words])
    return text


df['content'] = df['content'].apply(clean_text)
```

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet


# Download necessary NLTK resources
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

stemmer = PorterStemmer()


def apply_stemming(text):
    words = word_tokenize(text)  # Tokenize words
    stemmed_words = [stemmer.stem(word) for word in words]  # Apply stemming
    return ' '.join(stemmed_words)


# Example usage
df['content'] = df['content'].apply(apply_stemming)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000)  # Keep top 5000 words
X_tfidf = tfidf.fit_transform(df['content'])

print(X_tfidf.shape)  # Check dimensions
```

```python
df
```

```python
x = X_tfidf
y = df['sentiment']
```

```python
x[0]
```

```python
y[0]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=.2, random_state=42)
```

```python
print(x_test.shape, y_test.shape)
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Initialize Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(x_train, y_train)

# Predict on the test set
y_pred = model.predict(x_test)
```

```python
# Calculate Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")

# Display Confusion Matrix
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Display Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred))
```
